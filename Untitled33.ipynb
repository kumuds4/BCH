## Final Consolidated Python Script for GitHub

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import TensorDataset, DataLoader

# --- 1. BiLSTM_OutagePredictor Class Definition ---
class BiLSTM_OutagePredictor(nn.Module):
    def __init__(self, input_dim=7, hidden_dim=64, num_layers=2, dense_dim=32):
        super(BiLSTM_OutagePredictor, self).__init__()

        # Bi-LSTM Layer for temporal channel feature extraction
        self.bilstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True
        )

        # Fully-Connected DNN Head for outage probability estimation
        self.dnn = nn.Sequential(
            nn.Linear(hidden_dim * 2, dense_dim),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(dense_dim, dense_dim // 2),
            nn.ReLU(),
            nn.Linear(dense_dim // 2, 1),
            nn.Sigmoid() # Outputs predicted outage probability P_out in [0, 1]
        )

    def forward(self, x):
        # Input shape x: (batch_size, sequence_length, input_dim)
        lstm_out, _ = self.bilstm(x)
        # Extract representation from the final sequence step
        final_state = lstm_out[:, -1, :]
        # Pass through DNN classifier head
        outage_prob = self.dnn(final_state)
        return outage_prob

# --- 2. Synthetic Data Generation Function ---
def generate_synthetic_data(num_samples=1000, sequence_length=10, input_dim=7):
    snr_db = np.random.uniform(-10, 20, num_samples) # SNR in dB
    # Simulate a logistic-like decay for outage probability based on SNR
    simulated_p_out = 1 / (1 + np.exp(0.5 * snr_db - 2)) + np.random.normal(0, 0.05, num_samples)
    simulated_p_out = np.clip(simulated_p_out, 0.001, 0.999) # Clip to avoid 0 or 1 for BCE loss

    # Create input features (X): First feature is SNR, others are random noise
    X = np.random.randn(num_samples, sequence_length, input_dim)
    X[:, :, 0] = np.repeat(snr_db[:, np.newaxis], sequence_length, axis=1) # Set SNR feature

    # Target outage probabilities (y)
    y = simulated_p_out[:, np.newaxis]

    return torch.tensor(X, dtype=torch.float32), torch.tensor(y, dtype=torch.float32)

# --- 3. Generate Training and Testing Data ---
num_training_samples = 5000
X_train, y_train = generate_synthetic_data(num_training_samples)

num_test_samples = 1000
X_test, y_test = generate_synthetic_data(num_test_samples)

print(f"Generated X_train shape: {X_train.shape}")
print(f"Generated y_train shape: {y_train.shape}")
print(f"Generated X_test shape: {X_test.shape}")
print(f"Generated y_test shape: {y_test.shape}")

# Display a quick plot to verify the synthetic training data distribution
plt.figure(figsize=(8, 5))
plt.scatter(X_train[:, 0, 0].numpy(), y_train.numpy(), alpha=0.1, s=10, label='Synthetic Data Points')
plt.title('Synthetic Outage Probability vs. SNR (Training Data)')
plt.xlabel('SNR (dB)')
plt.ylabel('Outage Probability')
plt.grid(True)
plt.yscale('log')
plt.ylim(1, 1e-5)
plt.gca().invert_yaxis()
plt.legend()
plt.savefig('synthetic_outage_probability_vs_snr_training.png')
plt.show()

# --- 4. Model Instantiation, Loss, Optimizer, DataLoaders ---
model = BiLSTM_OutagePredictor(input_dim=7, hidden_dim=64, num_layers=2, dense_dim=32)
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

train_dataset = TensorDataset(X_train, y_train)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)

num_test_samples_per_epoch = 500
X_test_epoch, y_test_epoch = generate_synthetic_data(num_test_samples_per_epoch)
test_dataset_epoch = TensorDataset(X_test_epoch, y_test_epoch)
test_loader_epoch = DataLoader(test_dataset_epoch, batch_size=64, shuffle=False)

# --- 5. Training Loop ---
num_epochs = 20
losses = []
test_losses = []

print("Starting training...")
for epoch in range(num_epochs):
    # Training phase
    model.train()
    epoch_loss = 0
    for batch_X, batch_y in train_loader:
        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_epoch_loss = epoch_loss / len(train_loader)
    losses.append(avg_epoch_loss)

    # Evaluation phase on test set after each epoch
    model.eval()
    epoch_test_loss = 0
    with torch.no_grad():
        for batch_X_test, batch_y_test in test_loader_epoch:
            test_outputs = model(batch_X_test)
            test_loss = criterion(test_outputs, batch_y_test)
            epoch_test_loss += test_loss.item()

    avg_epoch_test_loss = epoch_test_loss / len(test_loader_epoch)
    test_losses.append(avg_epoch_test_loss)

    print(f'Epoch [{epoch+1}/{num_epochs}], Training Loss: {avg_epoch_loss:.4f}, Test Loss: {avg_epoch_test_loss:.4f}')
print("Training complete!")

# --- 6. Plotting Training and Test Loss Over Epochs ---
plt.figure(figsize=(10, 6))
plt.plot(losses, label='Training Loss', color='blue')
plt.plot(test_losses, label='Test Loss', color='red', linestyle='--')
plt.title('Training and Test Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.legend()
plt.savefig('training_and_test_loss_over_epochs.png')
plt.show()

# --- 7. Plotting Predicted Outage Probability vs. SNR (Trained Model) ---
snr_values_db_eval = np.linspace(-15, 25, 200)
outage_probabilities_vs_snr_trained = []

model.eval()
with torch.no_grad():
    for snr_db in snr_values_db_eval:
        eval_input = torch.zeros(1, 10, 7)
        eval_input[:, :, 0] = snr_db
        predicted_p_out_single_trained = model(eval_input)
        outage_probabilities_vs_snr_trained.append(predicted_p_out_single_trained.item())

plt.figure(figsize=(10, 6))
plt.plot(snr_values_db_eval, outage_probabilities_vs_snr_trained, color='blue', linewidth=2, label='Trained Model Prediction')
plt.title('Predicted Outage Probability vs. SNR (Trained Model)')
plt.xlabel('SNR (dB)')
plt.ylabel('Predicted Outage Probability')
plt.grid(True, which="both", ls="--", c='0.7')
plt.yscale('log')
plt.ylim(1, 1e-5)
plt.gca().invert_yaxis()
plt.legend()
plt.tight_layout()
plt.savefig('predicted_outage_probability_vs_snr_trained_model.png')
plt.show()

# --- 8. Model Evaluation on Test Set and Plotting Actual vs Predicted ---
model.eval()
with torch.no_grad():
    test_outputs = model(X_test)
    test_loss = criterion(test_outputs, y_test)
print(f"Test Loss on full test set: {test_loss.item():.4f}")

plt.figure(figsize=(10, 6))
plt.scatter(X_test[:, 0, 0].numpy(), y_test.numpy(), alpha=0.2, label='Actual Outage Probability')
plt.scatter(X_test[:, 0, 0].numpy(), test_outputs.numpy(), alpha=0.2, color='red', label='Predicted Outage Probability')
plt.title('Actual vs. Predicted Outage Probability on Test Data')
plt.xlabel('SNR (dB)')
plt.ylabel('Outage Probability')
plt.grid(True)
plt.legend()
plt.yscale('log')
plt.ylim(1, 1e-5)
plt.gca().invert_yaxis()
plt.savefig('actual_vs_predicted_outage_probability_test_data.png')
plt.show()

# --- 9. Saving Datasets and Model ---
torch.save(X_train, 'X_train.pt')
torch.save(y_train, 'y_train.pt')
print('Training data (X_train.pt, y_train.pt) saved successfully.')

torch.save(X_test, 'X_test.pt')
torch.save(y_test, 'y_test.pt')
print('Testing data (X_test.pt, y_test.pt) saved successfully.')

torch.save(model.state_dict(), 'bilstm_outage_predictor.pth')
print('Trained model (bilstm_outage_predictor.pth) saved successfully.')

print("All datasets, model, and plots have been saved.")
```
