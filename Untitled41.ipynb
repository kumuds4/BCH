```python
# ==============================================================================
# ALL-IN-ONE COLAB SCRIPT: ARIS BI-LSTM OUTAGE PREDICTION, XAI & DDPG CONTROL
# ==============================================================================

# 1. Install required packages
!pip install shap -q

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import shap
import matplotlib.pyplot as plt

# Reproducibility and device initialization
np.random.seed(42)
torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[+] Active Computation Device: {device}\n")

# ==============================================================================
# STEP 1: SYNTHETIC DATASET GENERATION (Physical Impairments & Channels)
# ==============================================================================
def generate_aris_dataset(num_samples=10000, seq_len=10, gamma_th=0.05):
    """
    Simulates dynamic SISO ARIS communication over Cascaded Rayleigh channels
    including von Mises phase noise, pointing jitter, and HPA back-off ratios.
    """
    # Cascaded Rayleigh channel power gains (|h1|^2, |h2|^2)
    h1_sq = np.random.exponential(scale=1.0, size=(num_samples, seq_len, 1))
    h2_sq = np.random.exponential(scale=1.0, size=(num_samples, seq_len, 1))

    # Operational parameters & physical hardware impairments
    P_t = np.random.uniform(20, 46, size=(num_samples, seq_len, 1))        # Transmit Power (dBm)
    kappa_phi = np.random.uniform(1, 10, size=(num_samples, seq_len, 1))   # Von Mises Phase Noise
    x_in = np.random.uniform(0.1, 1.0, size=(num_samples, seq_len, 1))      # HPA Back-off ratio
    sigma_tx = np.random.uniform(0.01, 0.1, size=(num_samples, seq_len, 1)) # Pointing jitter X
    sigma_ty = np.random.uniform(0.01, 0.1, size=(num_samples, seq_len, 1)) # Pointing jitter Y

    # Construct 7D Feature Matrix over sequence time steps
    X = np.concatenate([h1_sq, h2_sq, P_t, kappa_phi, x_in, sigma_tx, sigma_ty], axis=2)

    # Calculate Instantaneous Received SNR at final temporal step T_s
    eff_loss = x_in[:, -1, 0] * (1.0 / (1.0 + sigma_tx[:, -1, 0] + sigma_ty[:, -1, 0]))
    P_t_linear = 10 ** (P_t[:, -1, 0] / 10.0) / 1000.0
    snr_instantaneous = P_t_linear * h1_sq[:, -1, 0] * h2_sq[:, -1, 0] * eff_loss

    # Ground-truth binary target: 1 = Outage Event, 0 = Normal Operation
    y = (snr_instantaneous < gamma_th).astype(np.float32)

    return X, y

print("[+] Generating SISO ARIS propagation dataset...")
X_raw, y_raw = generate_aris_dataset(num_samples=10000)

# Train/Test Partitioning (80% Train, 20% Test)
split = int(0.8 * len(X_raw))
X_train = torch.tensor(X_raw[:split], dtype=torch.float32)
X_test = torch.tensor(X_raw[split:], dtype=torch.float32)
y_train = torch.tensor(y_raw[:split], dtype=torch.float32)
y_test = torch.tensor(y_raw[split:], dtype=torch.float32)

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)
print(f"[+] Dataset created: {X_train.shape[0]} training sequences, {X_test.shape[0]} test sequences.\n")

# ==============================================================================
# STEP 2: BI-LSTM NEURAL NETWORK ARCHITECTURE & TRAINING
# ==============================================================================
class BiLSTMOutagePredictor(nn.Module):
    def __init__(self, input_dim=7, hidden_dim=32, dense_dim=16):
        super(BiLSTMOutagePredictor, self).__init__()
        self.bilstm = nn.LSTM(input_dim, hidden_dim, batch_first=True, bidirectional=True)
        self.fc1 = nn.Linear(hidden_dim * 2, dense_dim)
        self.relu = nn.ReLU()
        self.bn = nn.BatchNorm1d(dense_dim)
        self.fc2 = nn.Linear(dense_dim, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        lstm_out, _ = self.bilstm(x)
        out = lstm_out[:, -1, :]  # Extract final sequence step features
        out = self.relu(self.bn(self.fc1(out)))
        out = self.sigmoid(self.fc2(out))
        return out # Removed .squeeze(-1) to maintain 2D output

model = BiLSTMOutagePredictor().to(device)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)

# Training Protocol
epochs = 12
model.train()
print("[+] Training Bi-LSTM Neural Decoder...")
for epoch in range(epochs):
    epoch_loss = 0.0
    for batch_X, batch_y in train_loader:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)

        optimizer.zero_grad()
        predictions = model(batch_X).squeeze(-1) # Squeeze here for loss calculation
        loss = criterion(predictions, batch_y)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * batch_X.size(0)

    epoch_loss /= len(train_loader.dataset)
    if (epoch + 1) % 3 == 0 or epoch == 0:
        print(f"    Epoch [{epoch+1:02d}/{epochs:02d}] - BCE Loss: {epoch_loss:.4f}")

print("\n[+] Bi-LSTM Training Completed!\n")

# ==============================================================================
# STEP 3: EXPLAINABLE AI (XAI) SHAP INTERPRETABILITY ANALYSIS
# ==============================================================================
print("[+] Running SHAP Explainable AI (XAI) feature analysis...")
model.eval()

background = X_train[:200].to(device)
test_samples = X_test[:100].to(device)

# Changed from DeepExplainer to GradientExplainer for better LSTM support
explainer = shap.GradientExplainer(model, background)
shap_values = explainer.shap_values(test_samples)

# Aggregate SHAP contributions across sequence steps for plotting
feature_names = ['|h1|^2', '|h2|^2', 'P_t', 'kappa_phi', 'x_in', 'sigma_theta_x', 'sigma_theta_y']
shap_aggregated = np.mean(shap_values, axis=1)
test_samples_avg = test_samples.mean(dim=1).cpu().numpy()

# Plot SHAP Feature Sensitivity Summary
plt.figure(figsize=(10, 5))
shap.summary_plot(shap_aggregated, test_samples_avg, feature_names=feature_names, show=False)
plt.title("SHAP Feature Importance for ARIS Outage Prediction", fontsize=13, pad=15)
plt.tight_layout()
plt.show()

# ==============================================================================
# STEP 4: DEEP REINFORCEMENT LEARNING (DDPG) PHASE CONTROL DEMO
# ==============================================================================
class DDPGActor(nn.Module):
    def __init__(self, state_dim=7, action_dim=16):
        super(DDPGActor, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 64),
            nn.ReLU(),
            nn.Linear(64, action_dim),
            nn.Tanh() # Scales continuous phase adjustments between [-pi, pi]
        )

    def forward(self, state):
        return self.net(state) * np.pi

N_elements = 16
actor = DDPGActor(state_dim=7, action_dim=N_elements).to(device)

# Inference sample for continuous real-time phase correction
state_sample = torch.tensor(X_raw[:1, -1, :], dtype=torch.float32).to(device)
phase_adjustments = actor(state_sample)

print("\n==================================================================")
print("[+] DDPG Control Layer Initialized Successfully!")
print("==================================================================")
print(f"Sample Channel/Impairment State (t=T_s):\n{state_sample.cpu().detach().numpy()[0]}")
print(f"\nDDPG Predicted Continuous Phase Shifts for {N_elements} ARIS Elements (Radians):\n{phase_adjustments.cpu().detach().numpy()[0]}")
print("==================================================================")
```

```python

# ==============================================================================
# ALL-IN-ONE COLAB SCRIPT: ARIS BI-LSTM OUTAGE PREDICTION, XAI & DDPG CONTROL
# ==============================================================================

# 1. Install required packages
!pip install shap -q

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import shap
import matplotlib.pyplot as plt

# Reproducibility and device initialization
np.random.seed(42)
torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[+] Active Computation Device: {device}\n")

# ==============================================================================
# STEP 1: SYNTHETIC DATASET GENERATION (Physical Impairments & Channels)
# ==============================================================================
def generate_aris_dataset(num_samples=10000, seq_len=10, gamma_th=0.05):
    """
    Simulates dynamic SISO ARIS communication over Cascaded Rayleigh channels
    including von Mises phase noise, pointing jitter, and HPA back-off ratios.
    """
    # Cascaded Rayleigh channel power gains (|h1|^2, |h2|^2)
    h1_sq = np.random.exponential(scale=1.0, size=(num_samples, seq_len, 1))
    h2_sq = np.random.exponential(scale=1.0, size=(num_samples, seq_len, 1))

    # Operational parameters & physical hardware impairments
    P_t = np.random.uniform(20, 46, size=(num_samples, seq_len, 1))        # Transmit Power (dBm)
    kappa_phi = np.random.uniform(1, 10, size=(num_samples, seq_len, 1))   # Von Mises Phase Noise
    x_in = np.random.uniform(0.1, 1.0, size=(num_samples, seq_len, 1))      # HPA Back-off ratio
    sigma_tx = np.random.uniform(0.01, 0.1, size=(num_samples, seq_len, 1)) # Pointing jitter X
    sigma_ty = np.random.uniform(0.01, 0.1, size=(num_samples, seq_len, 1)) # Pointing jitter Y

    # Construct 7D Feature Matrix over sequence time steps
    X = np.concatenate([h1_sq, h2_sq, P_t, kappa_phi, x_in, sigma_tx, sigma_ty], axis=2)

    # Calculate Instantaneous Received SNR at final temporal step T_s
    eff_loss = x_in[:, -1, 0] * (1.0 / (1.0 + sigma_tx[:, -1, 0] + sigma_ty[:, -1, 0]))
    P_t_linear = 10 ** (P_t[:, -1, 0] / 10.0) / 1000.0
    snr_instantaneous = P_t_linear * h1_sq[:, -1, 0] * h2_sq[:, -1, 0] * eff_loss

    # Ground-truth binary target: 1 = Outage Event, 0 = Normal Operation
    y = (snr_instantaneous < gamma_th).astype(np.float32)

    return X, y

print("[+] Generating SISO ARIS propagation dataset...")
X_raw, y_raw = generate_aris_dataset(num_samples=10000)

# Train/Test Partitioning (80% Train, 20% Test)
split = int(0.8 * len(X_raw))
X_train = torch.tensor(X_raw[:split], dtype=torch.float32)
X_test = torch.tensor(X_raw[split:], dtype=torch.float32)
y_train = torch.tensor(y_raw[:split], dtype=torch.float32)
y_test = torch.tensor(y_raw[split:], dtype=torch.float32)

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)
print(f"[+] Dataset created: {X_train.shape[0]} training sequences, {X_test.shape[0]} test sequences.\n")

# ==============================================================================
# STEP 2: BI-LSTM NEURAL NETWORK ARCHITECTURE & TRAINING
# ==============================================================================
class BiLSTMOutagePredictor(nn.Module):
    def __init__(self, input_dim=7, hidden_dim=32, dense_dim=16):
        super(BiLSTMOutagePredictor, self).__init__()
        self.bilstm = nn.LSTM(input_dim, hidden_dim, batch_first=True, bidirectional=True)
        self.fc1 = nn.Linear(hidden_dim * 2, dense_dim)
        self.relu = nn.ReLU()
        self.bn = nn.BatchNorm1d(dense_dim)
        self.fc2 = nn.Linear(dense_dim, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        lstm_out, _ = self.bilstm(x)
        out = lstm_out[:, -1, :]  # Extract final sequence step features
        out = self.relu(self.bn(self.fc1(out)))
        out = self.sigmoid(self.fc2(out))
        return out.squeeze(-1)

model = BiLSTMOutagePredictor().to(device)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)

# Training Protocol (Updated to 50 Epochs)
epochs = 50
model.train()
print(f"[+] Training Bi-LSTM Neural Decoder for {epochs} Epochs...")
for epoch in range(epochs):
    epoch_loss = 0.0
    for batch_X, batch_y in train_loader:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)

        optimizer.zero_grad()
        predictions = model(batch_X)
        loss = criterion(predictions, batch_y)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * batch_X.size(0)

    epoch_loss /= len(train_loader.dataset)
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"    Epoch [{epoch+1:02d}/{epochs:02d}] - BCE Loss: {epoch_loss:.4f}")

print("\n[+] Bi-LSTM Training Completed!\n")

# ==============================================================================
# STEP 3: EXPLAINABLE AI (XAI) SHAP INTERPRETABILITY ANALYSIS
# ==============================================================================
print("[+] Running SHAP Explainable AI (XAI) feature analysis...")
model.eval()

# To prevent SHAP dimension bugs, create a wrapper model that handles single time steps or sequences
background = X_train[:100].to(device)
test_samples = X_test[:50].to(device)

try:
    explainer = shap.DeepExplainer(model, background)
    shap_vals = explainer.shap_values(test_samples)

    # Convert shap values tensor/list to numpy safely
    if isinstance(shap_vals, list):
        shap_vals_np = shap_vals[0]
    else:
        shap_vals_np = shap_vals

    # Mean aggregation across the temporal dimension
    shap_aggregated = np.mean(shap_vals_np, axis=1)
    test_samples_avg = test_samples.mean(dim=1).cpu().numpy()

    feature_names = ['|h1|^2', '|h2|^2', 'P_t', 'kappa_phi', 'x_in', 'sigma_theta_x', 'sigma_theta_y']

    # Plot SHAP Feature Sensitivity Summary
    plt.figure(figsize=(10, 5))
    shap.summary_plot(shap_aggregated, test_samples_avg, feature_names=feature_names, show=False)
    plt.title("SHAP Feature Importance for ARIS Outage Prediction", fontsize=13, pad=15)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"[!] SHAP Visualizer Note: {e}")
    print("[+] Falling back to Gradient-based feature sensitivity calculation...")

    # Fallback Gradient-based Sensitivity
    test_samples.requires_grad = True
    outputs = model(test_samples)
    outputs.sum().backward()
    gradients = test_samples.grad.abs().mean(dim=(0, 1)).cpu().numpy()

    feature_names = ['|h1|^2', '|h2|^2', 'P_t', 'kappa_phi', 'x_in', 'sigma_theta_x', 'sigma_theta_y']
    plt.figure(figsize=(9, 4.5))
    plt.barh(feature_names, gradients, color='skyblue')
    plt.xlabel("Mean Absolute Gradient (Feature Sensitivity)")
    plt.title("XAI Feature Importance via Model Gradients")
    plt.tight_layout()
    plt.show()

# ==============================================================================
# STEP 4: DEEP REINFORCEMENT LEARNING (DDPG) PHASE CONTROL DEMO
# ==============================================================================
class DDPGActor(nn.Module):
    def __init__(self, state_dim=7, action_dim=16):
        super(DDPGActor, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(state_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 64),
            nn.ReLU(),
            nn.Linear(64, action_dim),
            nn.Tanh() # Scales continuous phase adjustments between [-pi, pi]
        )

    def forward(self, state):
        return self.net(state) * np.pi

N_elements = 16
actor = DDPGActor(state_dim=7, action_dim=N_elements).to(device)

# Inference sample for continuous real-time phase correction
state_sample = torch.tensor(X_raw[:1, -1, :], dtype=torch.float32).to(device)
phase_adjustments = actor(state_sample)

print("\n==================================================================")
print("[+] DDPG Control Layer Initialized Successfully!")
print("==================================================================")
print(f"Sample Channel/Impairment State (t=T_s):\n{state_sample.cpu().detach().numpy()[0]}")
print(f"\nDDPG Predicted Continuous Phase Shifts for {N_elements} ARIS Elements (Radians):\n{phase_adjustments.cpu().detach().numpy()[0]}")
print("==================================================================")
```
